```python
class Pair:
    def __init__(self, first, second):
        if not isinstance(first, (int, float)) or not isinstance(second, (int, float)):
            raise ValueError("Поля должны быть числами.")
        if first > second:
            raise ValueError("Левая граница не может быть больше правой.")
        self.first = first
        self.second = second

    def read(self):
        f = float(input("first: "))
        s = float(input("second: "))
        self.__init__(f, s)

    def display(self):
        print(f"first = {self.first}, second = {self.second}")

    def rangecheck(self, number):
        return self.first <= number <= self.second


def make_pair(cls, *args):
    try:
        return cls(*args)
    except Exception as e:
        print(f"Ошибка: {e}")
        raise SystemExit(1)


if __name__ == '__main__':
    p = make_pair(Pair, 5.0, 15.0)
    p.display()
    print(p.rangecheck(10))
    print(p.rangecheck(20))

    p2 = make_pair(Pair, 0, 0)
    p2.read()
    p2.display()

    make_pair(Pair, 100, 50)
```
