```python
from datetime import datetime, timedelta

class Date:
    def __init__(self, *args):
        """Метод инициализации. Поддерживает создание из 3 чисел или из строки."""
        if len(args) == 3:

            self._date = datetime(args[0], args[1], args[2])
        elif len(args) == 1 and isinstance(args[0], str):

            self._date = datetime.strptime(args[0], "%Y.%m.%d")
        else:
            raise ValueError("Ожидается 3 числа (год, месяц, день) или строка формата 'YYYY.MM.DD'")


    def read(self):
        """Ввод с клавиатуры"""
        date_str = input("Введите дату в формате 'год.месяц.день' (например, 2004.08.31): ")
        self._date = datetime.strptime(date_str, "%Y.%m.%d")

    def display(self):
        """Вывод на экран в заданном формате"""
        print(self._date.strftime("%Y.%m.%d"))


    def add_days(self, days):
        """Вычисление даты через заданное количество дней"""
        return Date((self._date + timedelta(days=days)).year,
                    (self._date + timedelta(days=days)).month,
                    (self._date + timedelta(days=days)).day)

    def subtract_days(self, days):
        """Вычитание заданного количества дней из даты"""
        return self.add_days(-days)

    def is_leap_year(self):
        """Определение високосности года"""
        year = self._date.year
        return year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)

    def get_year(self): return self._date.year
    def get_month(self): return self._date.month
    def get_day(self): return self._date.day

    def set_year(self, year): self._date = self._date.replace(year=year)
    def set_month(self, month): self._date = self._date.replace(month=month)
    def set_day(self, day): self._date = self._date.replace(day=day)

    def __eq__(self, other): return self._date == other._date
    def __lt__(self, other): return self._date < other._date
    def __gt__(self, other): return self._date > other._date

    def days_between(self, other):
        """Вычисление количества дней между датами"""
        return abs((self._date - other._date).days)


def make_date(*args):
    """Внешняя функция для создания объекта с проверкой ошибок"""
    try:
        return Date(*args)
    except Exception as e:
        print(f"Ошибка: {e}")
        return None



if __name__ == '__main__':
    print("--- Демонстрация работы класса Date ---")


    d1 = make_date(2004, 8, 31)
    d2 = make_date("2023.01.01")
    print("Дата 1:", end=" "); d1.display()
    print("Дата 2:", end=" "); d2.display()


    print(f"\nГод {d1.get_year()} високосный? -> {d1.is_leap_year()}")
    print(f"Год {d2.get_year()} високосный? -> {d2.is_leap_year()}")


    print("\nПрибавляем 10 дней к Дате 1:")
    d1.add_days(10).display()

    print("Вычитаем 1 день из Даты 2:")
    d2.subtract_days(1).display()


    print(f"\nДней между 2004.08.31 и 2023.01.01: {d1.days_between(d2)}")


    print(f"Дата 1 > Дата 2? -> {d1 > d2}")
    print(f"Дата 1 == Дата 2? -> {d1 == d2}")


    print("\nМеняем год у Даты 2 на 2024:")
    d2.set_year(2024)
    d2.display()


    print("\nПроверка ввода с клавиатуры (read):")
    d3 = make_date(1, 1, 1) # создаем заглушку
    d3.read()
    print("Вы ввели:", end=" "); d3.display()
```
